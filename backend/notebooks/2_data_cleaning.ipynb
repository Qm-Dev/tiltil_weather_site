This next block describes each of the columns previously obtain on the code block (backed up by the Weatherlink documentation available on the Windows program):
* **record_date:** Date of the archive. Follows a "DD-MM-YY HH:MM:SS" format. The measurement started in January 8, 2008 at 14:14:00.
* **avg_temp:** Average temperature measured during the interval. Reported in Celsius (°C).
* **hi_temp:** Highest temperature measured during the interval. Reported in Celsius (°C).
* **low_temp:** Lowest temperature measured during the interval. Reported in Celsius (°C).
* **out_hum:** Relative humidity. This is the relation between the amount of water vapor present in the air and the maximum amount of water vapor the air can contain at a specific temperature, expressed in percentage (%).
* **dew_pt:** Temperature the air needs to be cooled to in order to produce a relative humidity of 100%. Reported in Celsius (°C).
* **wind_speed:** Average wind speed reading from the station for the interval. Reported in km/h.
* **wind_direction:** Wind Direction. Composed of 16 "bins" representing the 16 compass points, the dominant wind direction (and therefore, the one reported) will be the one that has the most "markers" registered during the time of archive. Usually, it can take the value `---` if there is no wind speed, but can also indicate **bad or missing data.**
* **wind_run:** "Amount" of wind passing the station during a given period of time. Calculated by multiplying Avg. Wind Speed for each archive record by the archive interval. Reported in kilometers.
* **hi_speed:** Highest wind speed value recorded during the interval. Reported in km/h.
* **hi_dir:** Wind Direction of the highest wind speed value recorded.
* **wind_chill:** Perception of air temperature, taking into account the speed of the wind. Calculated with the Osczevski (1995) equation and reported in Celsius (°C)
* **heat_index:** Also known as *apparent temperature*, is how temperature feels like to the human body when relative humidity is combined with the air temperature. Reported in Celsius (°C).
* **thw_index:** Includes the effects of the wind into the Heat Index. Reported in Celsius (°C).
* **thsw_index:** Includes the effects of solar radiation into the THW Index. Reported in Celsius (°C).
* **bar:** Air pressure. Reported in mbar (1 mbar is equivalent to 1 hPa).
* **rain:** Amount of rain that has fallen during the archive interval. Reported in millimeters (mm).
* **rain_rate:** Amount of rain that would fall over a time period. Reported in mm/hr.
* **solar_rad:** Intensity of the sun's radiation reaching a horizontal surface. The value, reported in Watts per square meter (W/m2), is the average solar radiation measured over the archive interval.
* **solar_energy:** Amount of accumulated solar radiation energy over a period of time. Reported in Langleys (L); 1 langley is equivalent to 41840 joules per square metre (J/m2).
* **hi_solar_rad:** Peak solar radiation measured during the archive interval.
* **uv_index:** Intensity of UV, using a scale of 0 to 16. The value logged by WeatherLink is the average UV measured during the archive interval.

# Data Cleaning

## Duplicates

```python
df.duplicated().sum()
```

## Nulls

```python
display(df.isnull().sum())
```

```python
print(f"Total null amount: {df.isnull().sum().sum()}")
```

```python
for column in df:
    if df[column].isnull().sum()/df[column].notnull().sum() != 0:  
        print(f"{column}: {df[column].isnull().sum()/df[column].notnull().sum():.5f} ({df[column].isnull().sum()})")
```

Wind Direction and High Direction are the two columns with the most amount of null values. This makes sense since the Weatherlink program treats the no wind periods as "bad data" (null).

```python
df[(df["wind_direction"].isna()) | (df["hi_dir"].isna())]
```

## Removing unreliable data
According to the Weatherlink program documents, `ISS Recept` represents the overall quality level of the radio reception between the Vantage Pro consoles and the ISS transmitter. Reported as a number (percentage), the closer to `100`, the more reliable the data reported during the interval is; on the other hand, the closer to `0`, the more unreliable the data reported is.

```python
display(pd.DataFrame(df["iss_recept"].value_counts().sort_index(ascending=False)).head(10))
print(f"Total rows: {len(df)}")
```

Almost 99% of rows report an `ISS Recept` of 99% or greater. We will work with those records.

```python
df_copy = df.copy()
```

```python
df_copy = df_copy[df_copy["iss_recept"] >= 99]
df_copy
```

Checking the nulls again:

```python
for column in df_copy:
    if df_copy[column].isnull().sum()/df_copy[column].notnull().sum() != 0:  
        print(f"{column}: {df_copy[column].isnull().sum()/df_copy[column].notnull().sum():.5f} ({df_copy[column].isnull().sum()})")

print(f"\nTotal null amount: {df_copy.isnull().sum().sum()}")
```

Almost all columns got cleaned of nulls with this step. Those remaining with nulls are Wind related, THSW index and UV related measurements.

## Cleaning Wind measurements
### Calm (No Wind)

```python
no_wind_filter = (df_copy["wind_direction"].isna()) & (df_copy["wind_speed"] == 0) & (df_copy["wind_run"] == 0) & (df_copy["hi_speed"] == 0) & (df_copy["hi_dir"].isna())
df_copy[no_wind_filter][["record_date", "wind_speed", "wind_direction", "wind_run", "hi_speed", "hi_dir"]]
```

```python
df_copy[["wind_direction", "hi_dir"]] = df_copy[["wind_direction", "hi_dir"]].replace(np.nan, "Calm")
df_copy[no_wind_filter][["record_date", "wind_speed", "wind_direction", "wind_run", "hi_speed", "hi_dir"]]
```

```python
for column in df_copy:
    if df_copy[column].isnull().sum()/df_copy[column].notnull().sum() != 0:  
        print(f"{column}: {df_copy[column].isnull().sum()/df_copy[column].notnull().sum():.5f} ({df_copy[column].isnull().sum()})")

print(f"\nTotal null amount: {df_copy.isnull().sum().sum()}")
```

## Dropping THSW Index and UV related measurements

**THSW Index:** Calculation of this index is unknown.

**UV:** Redundancy.

```python
df_copy = df_copy.drop(columns=["thsw_index", "uv_index", "uv_dose", "hi_uv"])
df_copy
```

```python
print(f"Total null amount: {df_copy.isnull().sum().sum()}")
```

## Archive Interval
We will also keep the same archive interval for the records.

```python
df_copy = df_copy[df_copy["arc_int"] == 15].reset_index(drop=True)
df_copy
```
